```python
# Here we temperaly use w1, which actually is w1/sigma
def cal(n, w1, w2, w3, pi, xi_out, sigma):
    sending_rate = -1+w1*(xi_out)/(w2*n*sigma+w3*(pi)*(xi_out))
    print(w1*(xi_out))
    print(w2*n*sigma)
    print(w3*(pi)*(xi_out))
    return sending_rate * sigma
```

```python
n = 4
w1, w2, w3 = 15, 10, 1
xi_out = 12.8
sigma = 0.95
for pi in [0.5,1.5,3.5,-0.5]:
    print(cal(n, w1, w2, w3, pi, xi_out, sigma))
```

```python
for pi in [0.0,1.5,2.0]:
    print(cal(2, w1, w2, 1, pi, 12.8, sigma))
```
