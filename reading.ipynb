```python
import json
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python

# 设置 Seaborn 样式（可选）
sns.set(style="darkgrid")

# 定义 JSON 文件路径
json_file_path = r'D:\learning\reinforcement_learning\light_mappo\results\MyEnv\MyEnv\rmappo\check\run3\logs\summary.json'

# 检查文件是否存在
if not os.path.exists(json_file_path):
    raise FileNotFoundError(f"JSON 文件未找到，请检查路径: {json_file_path}")

# 读取 JSON 文件
with open(json_file_path, 'r') as f:
    summary = json.load(f)

# 初始化存储字典
loss_metrics = {
    'value_loss': [],
    'critic_grad_norm': []
}

# 遍历 JSON 数据，提取 'value_loss' 和 'critic_grad_norm' 指标
for key, data in summary.items():
    # 提取 'value_loss' 数据
    if key.endswith('logs/value_loss/value_loss') or 'value_loss' in key:
        for entry in data:
            timestamp, step, value = entry
            loss_metrics['value_loss'].append((step, value))
    
    # 提取 'critic_grad_norm' 数据
    elif key.endswith('logs/critic_grad_norm/critic_grad_norm') or 'critic_grad_norm' in key:
        for entry in data:
            timestamp, step, value = entry
            loss_metrics['critic_grad_norm'].append((step, value))

# 检查是否提取到 'value_loss' 数据
if not loss_metrics['value_loss']:
    raise ValueError("未能在 JSON 文件中找到 'value_loss' 指标的数据。请检查 JSON 文件内容。")

# 检查是否提取到 'critic_grad_norm' 数据
if not loss_metrics['critic_grad_norm']:
    raise ValueError("未能在 JSON 文件中找到 'critic_grad_norm' 指标的数据。请检查 JSON 文件内容。")

# 按步数排序
loss_metrics['value_loss'].sort(key=lambda x: x[0])
loss_metrics['critic_grad_norm'].sort(key=lambda x: x[0])

# 分别提取步数和对应的 loss 值
steps_value_loss, values_value_loss = zip(*loss_metrics['value_loss'])
steps_critic_grad_norm, values_critic_grad_norm = zip(*loss_metrics['critic_grad_norm'])

# 打印提取的数据（可选）
print("提取的 'value_loss' 数据:")
for step, value in loss_metrics['value_loss']:
    print(f"步数: {step}, Value Loss: {value}")

print("\n提取的 'critic_grad_norm' 数据:")
for step, value in loss_metrics['critic_grad_norm']:
    print(f"步数: {step}, Critic Grad Norm: {value}")

# 绘制 Value Loss 曲线
plt.figure(figsize=(12, 6))
plt.plot(steps_value_loss, values_value_loss, label='Value Loss', color='blue', linewidth=2)
plt.title('训练过程中的 Value Loss 变化', fontsize=16)
plt.xlabel('步数 (Steps)', fontsize=14)
plt.ylabel('Value Loss', fontsize=14)
plt.legend(fontsize=12)
plt.grid(True)

# 保存 Value Loss 图表为 PNG 文件
value_loss_image_path = r'D:\learning\reinforcement_learning\light_mappo\results\MyEnv\MyEnv\rmappo\check\run3\logs\value_loss_plot.png'
plt.savefig(value_loss_image_path, dpi=300)
print(f"Value Loss 图表已保存至: {value_loss_image_path}")

# 显示 Value Loss 图形
plt.show()

# 绘制 Critic Grad Norm 曲线
plt.figure(figsize=(12, 6))
plt.plot(steps_critic_grad_norm, values_critic_grad_norm, label='Critic Grad Norm', color='green', linewidth=2)
plt.title('训练过程中的 Critic Grad Norm 变化', fontsize=16)
plt.xlabel('步数 (Steps)', fontsize=14)
plt.ylabel('Critic Grad Norm', fontsize=14)
plt.legend(fontsize=12)
plt.grid(True)

# 保存 Critic Grad Norm 图表为 PNG 文件
critic_grad_norm_image_path = r'D:\learning\reinforcement_learning\light_mappo\results\MyEnv\MyEnv\rmappo\check\run3\logs\critic_grad_norm_plot.png'
plt.savefig(critic_grad_norm_image_path, dpi=300)
print(f"Critic Grad Norm 图表已保存至: {critic_grad_norm_image_path}")

# 显示 Critic Grad Norm 图形
plt.show()
```
